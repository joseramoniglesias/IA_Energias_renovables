```python
from openai import OpenAI

client = OpenAI(base_url="http://localhost:1234/v1", api_key="lm-studio",timeout=600)

completion = client.chat.completions.create(
  #model="lmstudio-ai/gemma-2b-it-GGUF", # this field is currently unused
  model = "local-mode",
  messages=[
    {"role": "system", "content": "Eres un asistente en español y ayudas con respuestas breves."},
    {"role": "user", "content": "Buenos días, ¿cuantos días llueve en Colombia al año?"}
  ],
  temperature=0.7,
)

msg = completion.choices[0].message

print(msg.content)
```
